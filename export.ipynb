```python

import subprocess
import shutil
from pathlib import Path
from IPython import get_ipython

notebook = Path("part1/tutorial.ipynb").resolve()
html_file = notebook.with_suffix(".html")
pdf_file = notebook.with_suffix(".pdf")

# Convert notebook to HTML
subprocess.run(
    [
        "python", "-m", "jupyter", "nbconvert",
        "--to", "html", str(notebook)
    ],
    check=True,
)

# Find Microsoft Edge
edge = (
    shutil.which("msedge")
    or r"C:\Program Files (x86)\Microsoft\Edge\Application\msedge.exe"
)

if not Path(edge).is_file():
    edge = r"C:\Program Files\Microsoft\Edge\Application\msedge.exe"

if not Path(edge).is_file():
    raise FileNotFoundError("Microsoft Edge was not found.")

# Print HTML directly to PDF
subprocess.run(
    [
        edge,
        "--headless",
        "--disable-gpu",
        "--no-pdf-header-footer",
        f"--print-to-pdf={pdf_file}",
        html_file.as_uri(),
    ],
    check=True,
    stdout=subprocess.DEVNULL,
    stderr=subprocess.DEVNULL,
)

print(f"PDF saved to: {pdf_file}")
```
